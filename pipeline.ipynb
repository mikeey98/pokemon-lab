Some names were saved with the wrong text encoding in the original file,
e.g. `Nidoranâ™€` instead of `Nidoran♀`. Encoding them back to cp1252 and
decoding as UTF-8 restores the original characters.

```python
broken = pokemon["Name"].str.contains("Ã|â", regex=True)
pokemon.loc[broken, "Name"] = pokemon.loc[broken, "Name"].map(
    lambda name: name.encode("cp1252").decode("utf-8")
)

pokemon.loc[broken, ["Num", "Name"]]
```

```python
def find_twins(name, n=5):
    selected = pokemon.loc[pokemon["Name"] == name]
    chosen_num = selected["Num"].iloc[0]
    chosen_stats = selected[stats].iloc[0]

    others = pokemon.loc[pokemon["Num"] != chosen_num].copy()
    others["distance"] = (others[stats] - chosen_stats).abs().sum(axis=1)

    chosen_row = selected.iloc[[0]].copy()
    chosen_row["distance"] = 0

    result = pd.concat(
        [chosen_row, others.nsmallest(n, "distance")],
        ignore_index=True
    )
    return result[["Name", "Type1", "distance"] + stats]
```

```python
api_pokemon = pd.read_csv(
    "https://raw.githubusercontent.com/PokeAPI/pokeapi/master/data/v2/csv/pokemon.csv"
)

api_pokemon.loc[
    api_pokemon["identifier"].str.contains("-mega"),
    ["identifier", "id"],
].head(10)
```

```python
mega = pokemon.loc[
    pokemon["Name"].str.startswith("Mega "),
    ["Name", "Num"],
].copy()

def mega_identifier(name):
    words = name.removeprefix("Mega ").lower().split()
    if words[-1] in ("x", "y"):
        return f"{'-'.join(words[:-1])}-mega-{words[-1]}"
    return f"{'-'.join(words)}-mega"

mega["identifier"] = mega["Name"].map(mega_identifier)

mega_matches = mega.merge(
    api_pokemon[["species_id", "identifier", "id"]],
    left_on=["Num", "identifier"],
    right_on=["species_id", "identifier"],
    how="left",
    validate="one_to_one",
)

display(mega_matches[["Name", "identifier", "id"]])
print("Mega-former utan träff:", mega_matches["id"].isna().sum())
```

```python
assert mega_matches["id"].notna().all()

print("special_sprite_ids = {")
for _, row in mega_matches.iterrows():
    print(f'    "{row["Name"]}": {int(row["id"])},')
print("}")
```

```python
start = 50  # Ändra sedan till 10, 20, 30, 40 ...

for _, row in mega_matches.iloc[start:start + 10].iterrows():
    print(f'    "{row["Name"]}": {int(row["id"])},')
```

```python
non_mega = pokemon.loc[~pokemon["Name"].str.startswith("Mega ")]

other_forms = non_mega.loc[
    non_mega["Num"].duplicated(keep=False),
    ["Num", "Name"],
]

other_forms.head(20)
```

```python
api_pokemon.loc[
    api_pokemon["species_id"].isin([382, 383]),
    ["species_id", "identifier", "id"],
]
```

```python
def api_identifier(name):
    if name == "Hoopa Confined":
        return "hoopa"

    words = name.lower().split()

    if words[-1] in {"forme", "cloak", "mode", "size"}:
        words.pop()

    if words[0] in {"primal", "black", "white"}:
        words.append(words.pop(0))

    return "-".join(words)


form_matches = other_forms.assign(
    identifier=other_forms["Name"].map(api_identifier)
).merge(
    api_pokemon[["species_id", "identifier", "id"]],
    left_on=["Num", "identifier"],
    right_on=["species_id", "identifier"],
    how="left",
    validate="one_to_one",
)

print("Matchade:", form_matches["id"].notna().sum(), "av", len(form_matches))
display(form_matches.loc[
    form_matches["id"].isna(),
    ["Num", "Name", "identifier"],
])
```

```python
assert form_matches["id"].notna().all()

form_overrides = form_matches.loc[
    form_matches["id"].ne(form_matches["Num"]),
    ["Name", "id"],
]

print("ID:n att lägga till:", len(form_overrides))

start = 30  # Kör sedan med 10, 20 och 30
for _, row in form_overrides.iloc[start:start + 10].iterrows():
    print(f'    "{row["Name"]}": {int(row["id"])},')
```
